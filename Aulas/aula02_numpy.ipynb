```python
import numpy as np
lista_num = np.array([1, 2, 3, 4, 5, 6])
print(lista_num)

print("Array: " , lista_num)
print("Tipo: ", type(lista_num))
print("Dimensão: ", lista_num.ndim)
print("Tamanho: ", lista_num.size)
```

```python
a = np.array([1, 2, 3])
b = np.array([[1, 2, 3], [4 ,5, 6]])
zeros = np.zeros((2, 3))
ones = np.ones((3, 3))

seq = np.arange(0, 10, 2)
linspace = np.linspace(0 ,20 , 5)

print("a: \n", a)
print("b: \n", b)
print("zeros: \n", zeros)
print("ones: \n", ones)
print("seq: ", seq)
print("linspace:", linspace)
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 20, 30 , 40 , 50])
x1 = np.arange(0 , 50 , 10)
x2 = np.arange(0 , 20 , 2)
print(x1)
print(x2)
print("soma: ", x + y)
print("multiplicação: ", x * y)
print("quadrado: ", x ** 2)
print("seno: ", np.sin(x))
print("média: ", np.mean(x))
print("desvio padrão: ", np.std(x))
```

```python
Lista_notas = np.array([9.1, 3.5, 9.8, 6.4,7.8 ])
print("Maximo:", Lista_notas.max())
print("Minimo:", Lista_notas.min())
print("Média:", Lista_notas.mean())
print("Posição da menor nota:", Lista_notas.argmin())
```

```python
import numpy as np

d = np.arange(1 , 11, 2)
print("array:", d)

e = np.full(3 ,3)
print("matriz: \n" , e)

f = np.arange(0 ,21 ,2)
print("array:",f)


m = np.array([2 ,4 ,6 , 8])
n = np.array([1 ,3 ,5 , 7])

print("M + N:",m + n)
print("M X N:", m * n)
print("M ** 2:", m ** 2)

rand1 = np.random.rand(10)
print("Aleatorios entre 0 e 1", rand1)

rand2 = np.random.randint(1, 101, 5)
print("Aleatorios entre 1 e 100", rand2)
```

```python
matriz = np.arange(0, 20).reshape(4, 5)
print(matriz)

print(matriz.sum())

print(matriz.mean())

print(matriz[: , 0])

print(matriz[-1 , -1])
```

```python
dados = np.random.randint(0 ,51 ,100)

print(dados)
print(np.mean(dados))
print(np.median(dados))
print(np.std(dados))

contagem = np.bincount(dados)
moda = np.argmax(contagem)
print(moda)
```
