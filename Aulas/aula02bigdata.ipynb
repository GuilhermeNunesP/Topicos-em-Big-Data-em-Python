```python
base = int(input("Digite a base do retângulo: "))
altura = int(input("Digite a altura do retângulo: "))

areaR = base * altura
print(f"A area do retangulo é {areaR}")

lado = int(input("Digite o valor do lado do quadrado: "))
areaQ = lado * lado
print(f"A area do quadrado é {areaQ}")

custo_produto = float(input("Digite o valor do produto: "))
desconto_produto = int(input("Digite o desconto do produto: "))

produtocomdesconto = (custo_produto * desconto_produto ) / 100
valorproduto = custo_produto - produtocomdesconto
print(f"O valor do produto com desconto é {valorproduto}")

raio = float(input("Digite o valor do raio do circulo: "))
areaC = raio * 3,14
print(f"A area do circulo é {areaC}")

dinheiroemreal = float(input("Digite o valor que deseja converter em dólar: "))
dinheiroemdolar = dinheiroemreal / 5,19

print(f"O valor convertido em dólar é {dinheiroemdolar} ")

dinemdolar = float(input("Digite o valor que deseja converter em reais: "))
dinemreais = dinemdolar * 5

print(f"O valor convertido em reais é {dinemreais} ")


```

```python
login = "admin"
senha = "123"
login2 = "admin2"
senha2 = "987"

usuariologin = input("Digite seu login: ")
usuariosenha = input("Digite sua senha: ")

if usuariologin == login and usuariosenha == senha:
  print("Usuario1 acessado com sucesso ")
elif usuariologin == login2 and usuariosenha == senha2:
  print("Usuario2 acessado com sucesso ")
else:
  print("Login não existe ou a senha está incorreta")



```

```python
salario = float(input("Digite seu salario: "))

if salario <= 2000:
  print("Voce é estagiario!!")
elif salario > 2000 and salario <= 3500:
  print("Voce é Junior!!")
elif salario > 3500 and salario <= 6500:
  print("Voce é Pleno!!")
elif salario > 6500 and salario <= 15000:
  print("Voce é Senior!!")
else:
  print("Voce é gerente de projetos!")
```
