```python
df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv")
df.head()

print(df.info())
print(df.isnull().sum())

df.drop(columns=["deck", "embark_town"], inplace=True)

df["age"].fillna(df["age"].mean())

df.rename(columns={"sex": "Sexo", "age": "Idade"}, inplace=True)

```

```python
df.head(30)
```

```python
df.rename(columns={"sex": "Sexo", "age": "Idade"}, inplace=True)
```

```python
df["class"].value_counts().plot(kind="bar", title="Distribuição por Classe")
plt.show()

df["Idade"].plot(kind="hist", bins=20, color="skyblue", edgecolor="black")
plt.title("Distru=ibuição de Idades")
plt.xlabel("Idade")
plt.show()


sns.boxplot(x="class", y="fare", data=df)
plt.title("Distribuição de Tarifa por Classe")
plt.show()

```

```python
sns.scatterplot(x="age", y="fare", hue="sex", data=df)
plt.title("Idade x Tarifa")
plt.show()
```
