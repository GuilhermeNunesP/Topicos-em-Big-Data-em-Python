```python
import numpy
import pandas
import sklearn
import matplotlib
```

```python
import pandas as pd

dados = {
    "Nome": ["Ana","Pedro", "Lucas"],
    "Idade": [23, 30, 25],
    "Cidade": ["SP", "RJ", "BH"]

}
df = pd.DataFrame(dados)

print(df)

print(df[df["Idade"] > 24])
```

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print("Soma:", np.sum(arr))
print("Media:", np.mean(arr))
print("Desvio padrao:", np.std(arr))

matriz = np.array([[1, 2], [3, 4]])
print("Determinante:", np.linalg.det(matriz))
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

modelo = LinearRegression()
modelo.fit(X, y)

print("Previsao:", modelo.predict([[10]]))
```

```python
import matplotlib.pyplot as plt

anos = [2019, 2020, 2021, 2022, 2023]
vendas = [100, 120, 90, 150, 180]

plt.plot(anos, vendas, marker="o", color="blue")
plt.title("Vendas por ano")
plt.xlabel("Ano")
plt.ylabel("Vendas")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

produtos = ["Notebook", "Mouse", "Teclado", "Monitor"]
vendas = [30, 120, 80, 62]

plt.bar(produtos, vendas)
plt.title("Vendas de produtos de informatica")
plt.xlabel("Produtos")
plt.ylabel("Quantidade Vendidas")
plt.show()
```
