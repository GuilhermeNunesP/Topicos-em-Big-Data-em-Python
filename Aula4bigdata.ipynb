```python
s = pd.Series([10 , 20, 30, 40, 50])
print(s)

dados = {
    "Nome": ["Carlos", "Ana", "Joâo", "Guilherme"],
    "Idade": [21, 30, 19, 18],
    "Nota": [8.6, 3.4, 7.8, 9.8]
}
df = pd.DataFrame(dados)
print("Coluna Nome: \n", df["Nome"])
print("2 primeiras linhas: \n", df.head(2))
print("apenas ultima linha: \n", df.tail(1))
```

```python
print("média de idade: \n", df["Idade"].mean())
print("Maior nota: \n", df["Nota"].max())
print("Menor nota: \n", df["Nota"].min())
print("Maior idade: \n", df["Idade"].max())
print("Menor idade: \n", df["Idade"].min())
print(" Visão geral do DataFrame \n", df.describe())
```

```python
aprovados = df[df["Nota"] >= 8]
print("Alunos aprovados: \n", aprovados)

jovens = df[df["Idade"] < 30]
print("Alunos jovens: \n", jovens)
```

```python
df["Situação"] = ["Aprovado" if nota >= 7 else "Reprovado" for nota in df["Nota"]]
print(df)
```

```python
df_dec = df.sort_values(by="Nota", ascending=False)
print("Ordem decrescente \n", df_dec)

df_cre = df.sort_values(by="Nota", ascending=True)
print("Ordem crescente \n", df_cre)
```
